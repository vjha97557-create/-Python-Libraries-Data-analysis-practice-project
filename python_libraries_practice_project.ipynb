```python
# inter quartile range 
# IQR=q3-q1
# q3=var['col_name'].quantile(0.75)
# q1=var['col_name'].quantile(0.25)
```

```python
q3=titanic['age'].quantile(0.75)
q1=titanic['age'].quantile(0.25)
print(q1)
print(q3)
```

```python
# IQR
titanic['age'].describe()
```

```python
# IQR
IQR=q3-q1
print(IQR)
```

```python
upper=q3+0.5*IQR
lower=q1-0.5*IQR
print(upper)
print(lower)
```

```python
outliers=titanic[(titanic['age']>upper)|(titanic['age']<lower)]
outliers
```

```python
out=[]
for i in titanic['age']:
    if i>upper:
        out.append(upper)
    elif i<lower:
        out.append(lower)
    else:
        out.append(i)
titanic['age']=out

print(out)
```
